# 1-D SINGLE-SLIT DIFFRACTION 

```python
from matplotlib import pyplot as plt
import numpy as np
import scipy
```

```python
lam = 532*10**-9
k = 2*np.pi/lam
z = 0.01 #1 cm

a  = 50.*10**-6
x = np.linspace(-10*a,10*a,1000) #points along screen
dxprime = 1e-7
xprime = np.linspace(-a/2,a/2,int(a/dxprime +1)) #limits of integration
```

```python
E = np.zeros(len(x), dtype = np.complex128)

r = lambda x: np.sqrt((x - xprime)**2. + z**2.)

for i in range(0, len(E)):
    E[i] = np.trapz( np.exp(1j*k*(r(x[i]) - 0))/r(x[i]), xprime)

```

```python
I_num = np.abs(E)**2
I_num /= np.max(I_num)
```

$$ I = I_0 sinc^2 \left( \frac{k a x}{2 z}\right) = I_0 sinc^2 (\beta)

```python
beta = np.pi*a*x/(lam*z)
I_fraun = np.sinc(beta/np.pi)**2.
```

```python
plt.plot(x, I_num, label ='Numerical')
plt.plot(x, I_fraun, linestyle='dashed', label ='Fraunhofer')
plt.xlabel(r'$x$')
plt.ylabel(r'$I^2(\text{normalized})$')
plt.legend()
plt.title('Distance to Screen(m): '+str(z))
plt.show()
```
